```python
from faker import Faker
import pandas as pd
from random import randrange

fake = Faker()
output = [{
        "name": fake.name(),
        "address": fake.address(),
        "city": fake.city(),
        "state": fake.state(),
        "email": fake.email(),
        "age": 10 + randrange(70)
    } for _ in range(100)]
df = pd.DataFrame(output)
df
```

```python
import ipywidgets as widgets

#bins = st.sidebar.slider("Bins:", value=10, min_value=3, max_value=20, step=1)
bins = widgets.IntSlider(description='Bins:', value=10, min=3, max=20, step=1)
display(bins)
```

```python
import matplotlib.pyplot as plt

df.hist(column="age", bins=bins.value)
plt.show()
```
