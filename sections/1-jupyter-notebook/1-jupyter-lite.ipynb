#### Generate 20 random data points along 3 dimensions

```python
import numpy as np

x, y, scale = np.random.randn(3, 20)
x, y, scale
```

#### Map each onto a scatterplot we'll create with Matplotlib

```python
from matplotlib import pyplot

fig, ax = pyplot.subplots()
ax.scatter(x=x, y=y, c=scale, s=np.abs(scale)*500)
ax.set(title="Some random data created for Jupyter Notebook")
pyplot.show()
```
